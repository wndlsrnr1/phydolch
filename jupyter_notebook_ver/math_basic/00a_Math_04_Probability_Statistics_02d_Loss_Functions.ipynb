### 핵심 원칙

**손실 함수는 확률 분포의 가정에서 자연스럽게 유도됩니다!**

손실 함수는 그냥 만든 게 아니라, "데이터가 이런 확률 분포를 따른다"고 가정하고 그 가정에서 자연스럽게 나온 거예요.

### 1) 평균 제곱 오차 손실 (Mean Squared Error, MSE Loss) ↔ 정규 분포 (Normal Distribution)

**언제 쓰나요?** 입력 $x$가 주어졌을 때 실수 값 $y$를 예측하는 회귀(regression) 문제에서 가장 널리 쓰는 손실 함수예요.

#### Step 0: 가정부터 세팅
- 관측 쌍: $(x_i, y_i)$, $i = 1, \, \ldots, \, n$
- 모델: $\hat{y}_i = f(x_i; \theta)$ (신경망, 선형 회귀 등)
- 오차: $\epsilon_i = y_i - f(x_i; \theta)$
- 가정: $\epsilon_i \sim \mathcal{N}(0, \sigma^2)$ (평균 0, 분산 $\sigma^2$인 정규분포)

> **왜 정규분포?**
> - 많은 작은 교란의 합 → 중심극한정리에 의해 정규분포로 근사
> - 실제 측정 오차가 정규분포로 모델링되는 경우가 많음
> - 물론 항상 맞지는 않지만 "가장 먼저 시도해 볼 만한" 합리적 출발점이에요.

#### Step 1: 한 데이터 포인트의 우도

$$P(y_i \,|\, x_i, \theta) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(y_i - f(x_i; \theta))^2}{2\sigma^2}\right)$$

- $f(x_i; \theta)$가 예측한 값 근처일수록 확률(우도)이 커짐
- 멀리 떨어질수록 지수 함수가 급격히 감소

#### Step 2: 전체 데이터의 우도

데이터가 독립이라고 가정하면 곱으로 묶을 수 있어요.

$$L(\theta) = \prod_{i=1}^n P(y_i \,|\, x_i, \theta)$$

#### Step 3: 로그 우도 = 덧셈 형태

$$\ell(\theta) = \log L(\theta) = -\frac{n}{2} \log(2\pi\sigma^2) - \frac{1}{2\sigma^2} \sum_{i=1}^n (y_i - f(x_i; \theta))^2$$

- 첫 항은 상수 (최적화와 무관)
- 두 번째 항에 우리가 익숙한 "제곱 오차"가 등장합니다.

#### Step 4: 손실 함수 탄생

음의 로그 우도를 손실로 두면

$$\text{NLL}(\theta) = -\ell(\theta) = \frac{1}{2\sigma^2} \sum_{i=1}^n (y_i - f(x_i; \theta))^2 + \text{const}$$

상수를 떼고 $\frac{1}{n}$으로 평균을 내면 바로 MSE가 돼요.

$$\text{MSE}(\theta) = \frac{1}{n} \sum_{i=1}^n (y_i - f(x_i; \theta))^2$$

> **요약**
> - 정규분포 가정 → 제곱 오차가 자연스럽게 등장
> - 우도를 최대화하는 것 = MSE를 최소화하는 것
> - "예측이 실제값과 얼마나 떨어져 있는가"를 제곱으로 측정하는 이유가 통계적으로 정당화됩니다.
### 2) 교차 엔트로피 손실 (CrossEntropy Loss) ↔ 다항 분포 (Multinomial Distribution) + 정보 이론

**언제 쓰나요?** 여러 클래스 중 하나를 예측해야 하는 분류 문제에서 표준으로 쓰이는 손실 함수예요.

#### Step 0: 준비물 정리
- 정답 라벨: one-hot 벡터 $y_i = [0, \, 	ext{...}, \, 1, \, 	ext{...}, \, 0]$
- 모델 출력: 소프트맥스 확률 $\hat{p}_i = [\hat{p}_{i,1}, \ldots, \hat{p}_{i,K}]$, 합은 1
- $y_{i,k} = 1$이면 $k$번째 클래스가 정답이라는 뜻

#### Step 1: 정보 이론 관점

교차 엔트로피는 "정답 분포 $P$"를 "모델 분포 $Q$"로 코딩할 때 필요한 평균 정보량이에요.

$$H(P, Q) = -\sum_{k=1}^K y_{i,k} \log \hat{p}_{i,k}$$

- 정답 클래스 하나만 1이므로 실질적으로는 $-\log \hat{p}_{i,k^*}$만 남아요.
- $\hat{p}_{i,k^*}$가 커질수록(정답을 더 확신할수록) 손실이 작아져요.
- $H(P, Q) = H(P) + D_{KL}(P\|\|Q)$ → 정답 분포의 엔트로피는 고정이므로, 교차 엔트로피를 줄이는 것은 KL 발산을 줄이는 것과 동일합니다.

#### Step 2: MLE 관점

레이블이 "카테고리/다항 분포"를 따른다고 보면, 한 데이터의 로그 우도는

$$\log P(y_i \,|\, x_i, \theta) = \sum_{k=1}^K y_{i,k} \log \hat{p}_{i,k}$$

전체 데이터에 대한 음의 로그 우도는

$$-\sum_{i=1}^n \sum_{k=1}^K y_{i,k} \log \hat{p}_{i,k}$$

→ 바로 우리가 사용하는 CrossEntropy Loss 형태죠. MLE는 이 값을 최소화하면서 $\theta$를 찾습니다.

#### Step 3: 두 관점이 하나로
- 정보 이론: 실제 분포와 모델 분포의 차이(정보 손실)를 최소화
- 통계/MLE: 데이터가 나올 가능성이 최대가 되도록 파라미터 학습

결국 둘 다 같은 식으로 귀결되며, 손실이 작을수록 모델 확률 $\hat{p}$가 정답 분포 $y$에 가까워지는 것을 의미합니다.

> **이름의 의미**
> 실제 분포(정답)와 예측 분포가 서로 "교차"하는 방식으로 엔트로피를 측정하기 때문에 Cross + Entropy 라고 부릅니다.
### 핵심 통찰

**손실 함수는 확률 분포의 가정에서 자연스럽게 유도됩니다!**

**이게 왜 중요한가요?** 
손실 함수는 그냥 만든 게 아니라, "데이터가 이 확률 분포를 따른다"고 가정하고 그 가정에서 나온 거예요.

- **MSE**: 오차가 정규분포 (Normal Distribution)를 따른다고 가정 → 제곱 오차가 자연스럽게 나타남
- **CrossEntropy**: 레이블이 다항분포 (Multinomial Distribution)를 따른다고 가정 → 로그 확률이 자연스럽게 나타남

**따라서**:
- 회귀 문제에 MSE를 쓰는 건, 오차가 정규분포라는 합리적 가정 때문이에요
- 분류 문제에 CrossEntropy를 쓰는 건, 레이블이 다항분포라는 합리적 가정 때문이에요

### 배치 정규화 (Batch Normalization)와 정규 분포

**한 줄 소개**: 학습 중 계속 흔들리는 활성값의 분포를 매 미니배치마다 "평균 0, 분산 1"에 가깝게 재조정해 주는 기법이에요.

#### 1. 왜 필요한가? (Internal Covariate Shift)
- 이전 층의 파라미터가 업데이트되면, 다음 층이 받는 입력 분포가 달라짐
- 매번 새 분포에 적응해야 하니 학습이 불안정하고 느려짐
- 특히 활성화 함수가 포화되는 구간(sigmoid, tanh 등)에서 문제가 크게 나타나요

#### 2. 배치 정규화의 수식

미니배치 $B = \{x_1, \, \ldots, \, x_m\}$에 대해

$$\mu_B = \frac{1}{m} \sum_{i=1}^m x_i, \quad \sigma_B^2 = \frac{1}{m} \sum_{i=1}^m (x_i - \mu_B)^2$$

표준화한 값은

$$\hat{x}_i = \frac{x_i - \mu_B}{\sqrt{\sigma_B^2 + \epsilon}}$$

마지막으로 학습 가능한 스케일·시프트를 되돌려 붙입니다.

$$y_i = \gamma \hat{x}_i + \beta$$

- $\gamma$: 얼마나 크게 키울지 (scale)
- $\beta$: 어디로 이동시킬지 (shift)
- $\epsilon$: 0으로 나누는 사고를 막기 위한 아주 작은 값

#### 3. 통계적 해석
- $\hat{x}_i$는 평균 0, 분산 1에 가깝게 정규화된 값 → 표준 정규분포에 근사
- 이렇게 맞춰 놓으면 다음 층은 항상 비슷한 분포의 입력을 받기 때문에 학습이 안정적
- $\gamma$, $\beta$ 덕분에 모델이 원래 필요로 하는 분포 모양은 다시 복원할 수 있어요 (정보 손실 X)

#### 4. 학습/추론 시 통계량 사용법
- **학습 중**: 현재 미니배치에서 계산한 $\mu_B$, $\sigma_B^2$ 사용
- **추론 시**: 학습 과정에서 누적해 둔 이동 평균 $\mu_{\text{running}}$, $\sigma^2_{\text{running}}$ 사용
  - 예: $\mu_{\text{running}} \leftarrow \alpha\,\mu_{\text{running}} + (1-\alpha)\,\mu_B$
- 배치 크기가 1이어도 안정적으로 추론 가능

#### 5. 기대 효과
- 학습률을 더 높게 설정해도 수렴이 빠르고 안정적
- 초기화에 덜 민감, 그래디언트 소실/폭발 완화
- 결국 "각 층이 받는 입력을 표준 정규분포 근처로 유지"한다는 점에서 통계적 의미가 분명해집니다.
