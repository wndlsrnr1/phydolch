```python
# x²의 미분 증명
print("="*60)
print("f(x) = x²의 미분 증명")
print("="*60)

# 미분 정의 적용
print("\n[Step 1] 미분 정의 사용:")
print("f'(x) = lim(h→0) [f(x+h) - f(x)] / h")
print("f'(x) = lim(h→0) [(x+h)² - x²] / h")
print()

print("[Step 2] (x+h)² 전개:")
print("(x+h)² = x² + 2xh + h²")
print()

print("[Step 3] 분자 계산:")
print("[(x+h)² - x²] = [x² + 2xh + h² - x²]")
print("              = 2xh + h²")
print()

print("[Step 4] 전체 식 정리:")
print("f'(x) = lim(h→0) [2xh + h²] / h")
print("f'(x) = lim(h→0) [h(2x + h)] / h")
print("f'(x) = lim(h→0) [2x + h]")
print()

print("[Step 5] 극한 계산:")
print("h → 0이므로")
print("f'(x) = 2x + 0 = 2x")
print()

print("="*60)
print("결론: f(x) = x²의 미분은 f'(x) = 2x")
print("="*60)
```
