엔트로피

엔트로피란? 확률변수 X가 만들어 내는 결과들의 평균 정보량 또는 불확실성의 크기를 수치로 표현한 값. 결과를 맞히기 어렵게 만드는 분포일 수록 엔트로피가 커지고, 언제나 같은 결과가 나오는 분포일수록 엔트로피가 0에 가까워진다.

Step 0: 표현 정리부터!

X: 우리가 관심 있는 확률변수 (예: 동전을 던졌을 때의 앞/뒤)
xi: 확률변수 X가 가질수 있는 i번째 결과
pi = P(X = xi) i번째의 결과가 일어날 확률인 확률변수의 확률 i번째 결과가 일어날 확률
logb 밑이 b인 로그, 보통 밑을 2로 두면 단위가 bit, 자연 로그 e를 쓰면 nat이된다. 

평균 정보량이란 말을 수식으로 담아내기 위해, 한 사건이 실제로 발생했을때 느끼는 놀라움을 먼저 정의해야한다. 

Step1. 한 사건에서 얻는 정보량 (Information Content)

확률이 pi인 사건 xi가 실제로 일어나면 얻는 정보량을 다음처럼 정의한다.

I(xi)
xi가 일어날때의 정보량이란? 
xi가 일어날때의 확률의 log 그리고 -를 곱함
희귀한 사건(pi가 작을 때)이 일어나면 -logpi값이 커져서 더 큰 정보를 준다. 
항상 일어나는 사건(pi = 1)은 I(xi) = 0 이라서 새로 알게 된 정보가 없다.

왜 로그를 사용하는가? 

독립적인 사건들의 정보량을 합으로 표현하고 싶기 때문이다.
로그는 곱셈을 덧셈으로 바꾸어준다.
동전을 두번 던져 앞이 나올 확률

즉, 로그 덕분에 독립 시행의 정보량 함 = 전체 시행 정보량 이라는 멋진 성질을 얻을 수 있다.

Step 3: 평균 정보량 -> 엔트로피 정의

각 사건이 가지는 정보량에 그 사건이 실제로 일어날 확률 pi를 곱해 평균을 내면, 그 분포가 가진 전체적인 불확실성 (평균 놀라움)을 얻을 수 있다. 

각 사건이 가지는 정보량 I(xi)에 그 사건이 실제로 일어날 확률 pi를 곱해 평균을 내면, 그 분포가 가진 전체적인 불확실성(평균 놀라움)을 얻을 수 있음. 

그 분포가 가진 전체적인 불확실성 = 엔트로피
H(X) E[I(X)] 그 분포의 불확실성이란 그 정보량이 가진 값들의 평균값이다. 
xi 일때의 정보량 * 그것이 일어날 확률 을 곱한다.
