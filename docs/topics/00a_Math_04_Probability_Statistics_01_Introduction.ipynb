### Step 2: 왜 이 단원이 필요한가?
머신러닝은 본질적으로 **확률론적 모델링**입니다. 모델은 데이터의 확률 분포를 학습하고, 예측은 확률로 표현됩니다.

- **CrossEntropy Loss**: 왜 분류 문제에서 사용하는가? → 다항 분포 (Multinomial Distribution)와 MLE
- **MSE (Mean Squared Error) Loss**: 왜 회귀 문제에서 사용하는가? → 정규 분포 (Normal Distribution)와 MLE
- **정규화**: 왜 배치 정규화 (Batch Normalization)가 작동하는가? → 통계적 정규화
- **불확실성**: 모델의 예측이 얼마나 확실한가? → 확률 해석

이 단원을 이해하면 손실 함수가 단순한 수식이 아니라 **확률론적 의미**를 가진다는 것을 알게 됩니다.

## 1.1 확률이란 무엇인가?

### Step 0: 기호 정리
- 사건 $A$: 우리가 관찰하려는 결과 (예: "비가 온다")
- 확률 $P(A)$: 사건 $A$가 일어날 가능성 (0~1 사이)
- 확률변수 $X$: 확률 실험의 결과를 숫자로 표현한 함수

### Step 1: 생활 속 직관
- **확실성**: "내일 해가 뜬다" → $P(A) = 1$
- **불확실성**: "내일 비가 올 확률은 70%" → $P(A) = 0.7$
- **표현 방식**: 확률은 불확실성을 숫자로 정량화한 값

### Step 2: 머신러닝 관점
- **분류 모델**: "이 이미지가 고양이일 확률은 0.85"
- **불확실성 해석**: 확률이 높을수록 모델의 확신도가 높음
- **데이터 분포 학습**: 모델이 실제 데이터의 확률 분포를 근사하도록 파라미터를 조정

### Step 3: 손실 함수와의 연결
1. **CrossEntropy Loss**: 모델이 예측한 분포와 실제 정답 분포의 차이를 측정 (다항 분포 기반)
2. **MSE Loss**: 예측과 실제 값의 차이를 제곱해 평균 (정규 분포 기반)
3. **Softmax**: 선형 출력을 확률 분포로 변환하여 CrossEntropy와 함께 사용
