# 학습 로드맵 (Learning Roadmap)

## 전체 커리큘럼 개요

이 튜토리얼은 **이론부터 실전까지** 단계별로 학습할 수 있도록 구성되었습니다.

## 단원별 학습 순서


### Phase 1: 수학 기초 (00, 00a, 00b)

**목적**: PyTorch를 이해하기 위한 수학적 배경

1. **00. 선형대수 기초**
   - 벡터와 행렬의 기하학적 의미
   - 내적과 행렬곱 (Matrix Multiplication)
   - 선형 변환과 차원 변환
   - **연결**: 텐서 (Tensor) 연산 이해의 기초

2. **00a. 미적분 기초**
   - 미분 (Derivative)의 기하학적 의미
   - 기울기 (Gradient)와 편미분 (Partial Derivative)
   - 체인룰 (Chain Rule) ⭐
   - 경사하강법 (Gradient Descent)
   - **연결**: 자동 미분 (Autograd)의 핵심

3. **00b. 확률과 통계**
   - 확률 분포 (Probability Distribution)
   - 최대우도 추정 (Maximum Likelihood Estimation, MLE)
   - 손실 함수 (Loss Function)의 통계적 의미
   - **연결**: CrossEntropy, MSE (Mean Squared Error)의 의미

**핵심**: 이 단계에서 **"왜 이 수학이 필요한가?"**를 이해해야 합니다.


### Phase 2: PyTorch 기초 (01~03)

**목적**: PyTorch의 핵심 구조와 자동 미분

4. **01. 텐서 (Tensor) 기초**
   - 텐서의 개념 (벡터/행렬의 일반화)
   - 차원과 shape
   - **연결**: 수학의 벡터/행렬 → 텐서

5. **02. 텐서 연산**
   - 행렬곱 (Matrix Multiplication)과 브로드캐스팅 (Broadcasting)
   - 인덱싱과 조작
   - **연결**: 선형 레이어 (Linear Layer) 구현의 기초

6. **03. 자동 미분 (Autograd)**
   - 자동 미분의 필요성
   - 계산 그래프 ⭐
   - 체인룰 (Chain Rule) 자동 적용
   - **연결**: 역전파 (Backpropagation)의 수학적 원리

**핵심**: 이 단계에서 **"PyTorch가 어떻게 작동하는가?"**를 이해합니다.


### Phase 3: 데이터와 신경망 (04~07)

**목적**: 실제 문제를 신경망으로 해결

7. **04. 데이터 준비**
   - 데이터셋 (Dataset)과 데이터로더 (DataLoader)
   - 전처리와 변환
   - **연결**: 실제 데이터를 텐서로

8. **05. 신경망 (Neural Network) 구조**
   - Layer → Module (모듈) → Model
   - nn.Module 상속
   - **연결**: 텐서 연산을 구조화

9. **06. 선형 레이어 (Linear Layer)**
   - 행렬곱 (Matrix Multiplication) + 편향 (Bias)
   - 내적의 의미
   - **연결**: 선형대수 + PyTorch 구현

10. **07. 합성곱 레이어 (Convolution Layer)**
    - 이미지 특화 레이어
    - 필터의 의미
    - **연결**: 지역 패턴 감지

**핵심**: 이 단계에서 **"신경망이 무엇을 하는가?"**를 이해합니다.


### Phase 4: 첫 프로젝트 (10)

**목적**: 전체 파이프라인 경험

11. **10. 첫 번째 프로젝트**
    - 데이터 → 모델 → 학습 → 평가
    - 디버깅
    - 실전 팁

**핵심**: 이 단계에서 **"실제로 작동하는 프로젝트"**를 완성합니다!


## 핵심 개념 연결도

```
수학 기초 (00, 00a, 00b)
     ↓
PyTorch 기초 (01~03)
     ↓
데이터 준비 (04)
     ↓
신경망 구조 (05~07)
     ↓
첫 프로젝트 (10)
```

**각 단계는 이전 단계를 기반으로 합니다!**


## 주요 연결고리

### 1. 체인룰 (Chain Rule) 연결
- **00a(미적분)**: 체인룰 이론
- **03(자동 미분, Autograd)**: 체인룰 자동 적용
- **역전파 (Backpropagation)**: 체인룰의 실전 구현

### 2. 행렬연산 연결
- **00(선형대수)**: 벡터/행렬 기하학
- **02(텐서 연산)**: 행렬곱 (Matrix Multiplication) 구현
- **06(선형 레이어, Linear Layer)**: 신경망의 기본 연산

### 3. 확률 연결
- **00b(확률통계)**: MLE 이론
- **손실함수 (Loss Function)**: CrossEntropy, MSE 구현
- **모델 출력**: 확률 해석

### 4. 데이터 연결
- **01(텐서)**: 모든 데이터는 텐서
- **04(데이터 준비)**: 실제 데이터 처리
- **10(프로젝트)**: 전체 파이프라인

**핵심**: 각 단원이 독립적이 아닌 **연결된 체계**입니다!


## 학습 방법

### 권장 순서
1. **순서대로** 학습 (건너뛰지 말 것)
2. **이전 단원 복습** 섹션 확인
3. **다음 단원 미리보기** 이해
4. **연습 문제** 풀기

### 각 단원에서 할 일

- [ ] 수학 기초 (공식 이해)
- [ ] PyTorch 코드 실행
- [ ] 시각화 확인
- [ ] 핵심 요약 복습

- 수학 기초 단원(00, 00a, 00b) 복습
