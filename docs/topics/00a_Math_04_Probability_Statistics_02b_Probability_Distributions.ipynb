# 2.2 확률 분포
---

### Step 0: 용어 미리보기
- **확률 분포**: 확률변수가 가질 수 있는 값과 그 확률을 전부 모아놓은 지도
- **PMF (Probability Mass Function)**: 이산 확률변수의 확률 분포
- **PDF (Probability Density Function)**: 연속 확률변수의 확률 분포

### Step 1: 왜 확률 분포를 공부할까?
- 데이터가 어떤 패턴을 따르는지 이해해야 모델이 제대로 학습
- 모델이 출력하는 값의 불확실성을 해석할 수 있음
- 손실 함수는 특정 분포를 가정하고 만들어지므로, 분포를 알아야 선택 이유를 이해할 수 있음

### Step 2: 큰 그림
- **이산 분포**: 셀 수 있는 값 (동전, 주사위, 분류 결과)
- **연속 분포**: 연속적인 값 (키, 온도, 노이즈)
- 각 분포마다 파라미터가 다르고, 머신러닝에서 활용되는 방식도 다름

### 이산 확률 분포 (Discrete Probability Distribution)

**이산이란?** 셀 수 있는 값들 (1, 2, 3, ... 또는 "고양이", "개", "새")

**확률 질량 함수 (Probability Mass Function, PMF)**: $P(X = x)$
- 각 값에 대해 정확히 그 값이 나올 확률

#### 1) 이항 분포 (Binomial Distribution)

##### Step 0: 기호 정리
- $n$: 반복 횟수 (시행 수)
- $k$: 성공 횟수
- $p$: 한 번 시행에서 성공할 확률
- $X$: 성공 횟수를 나타내는 확률변수

##### Step 1: 정의
- 동일한 베르누이 시행을 $n$번 반복할 때 성공 횟수의 분포
- 확률질량함수: $$P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}$$

##### Step 2: 공식 유도
1. **경우의 수**: $\binom{n}{k}$ (성공 위치 선택)
2. **각 경우의 확률**: $p^k (1-p)^{n-k}$ (독립 사건 곱셈 원리)
3. **모든 경우 합산**: 곱셈 결과에 경우의 수를 곱해 최종 확률 도출

##### Step 3: 예시 계산
- 공정한 동전($p=0.5$)을 $n=5$번 던질 때 $k=3$번 앞면이 나올 확률
- $P(X=3) = 10 \times (0.5)^3 \times (0.5)^2 = 0.3125$

##### Step 4: 기댓값과 분산
- $E[X] = np$
- $\text{Var}(X) = np(1-p)$
- 예: $n=100$, $p=0.5$ → $E[X]=50$, $\text{Var}(X)=25$

##### Step 5: 머신러닝 연결
- **이진 분류**에서 Positive(성공) vs Negative(실패) 횟수를 설명
- 로그우도(log-likelihood)를 최대화하면 이항 분포 기반의 CrossEntropy Loss와 연결
- 다항 분포는 이항 분포($k=2$)의 확장 버전

#### 2) 다항 분포 (Multinomial Distribution)

##### Step 0: 기호 정리
- $k$: 가능한 결과(클래스) 수
- $n$: 전체 시행 횟수
- $n_i$: 결과 $i$가 관측된 횟수
- $p_i$: 결과 $i$가 한 번의 시행에서 나타날 확률

##### Step 1: 정의
여러 개의 결과(outcome/category) 중 하나를 선택하는 확률 분포예요. 이항 분포를 여러 결과(outcome/category)로 확장한 형태입니다.

##### Step 2: 이항 분포와의 관계
- **이항 분포**: 2가지 결과(outcome) (성공/실패) → 다항 분포의 특수한 경우 ($k=2$)
- **다항 분포**: $k$개 결과(outcome/category) ($k \geq 2$)
- 이항 분포는 다항 분포에서 결과(outcome)가 2개일 때의 경우

##### Step 3: 직관과 예시
- 동전 던지기: 2가지 결과(outcome) (앞/뒤) → **이항 분포**
- 주사위 던지기: 6가지 결과(outcome) (1~6) → **다항 분포**
- 분류 문제: 여러 클래스 중 하나 선택 → **다항 분포**

##### Step 4: 확률질량함수
$n$번 시행에서 각 결과(outcome) $i$가 $n_i$번 나올 확률

$$P(X_1=n_1, X_2=n_2, \ldots, X_k=n_k) = \frac{n!}{n_1! n_2! \cdots n_k!} p_1^{n_1} p_2^{n_2} \cdots p_k^{n_k}$$

여기서:
- $n$: 전체 시행 횟수
- $n_i$: 결과(outcome) $i$가 나온 횟수 ($i = 1, 2, ..., k$)
- $p_i$: 한 번의 시행에서 결과(outcome) $i$가 나올 확률

##### Step 5: 제약 조건

**시행 횟수 제약**: 
$$\sum_{i=1}^k n_i = n$$

**의미**: 모든 결과(outcome)가 나온 횟수의 합은 전체 시행 횟수와 같아야 해요

**확률 제약**:
$$\sum_{i=1}^k p_i = 1$$

**의미**: 모든 결과(outcome)의 확률을 더하면 1이 되어야 해요 (어떤 결과(outcome)든 반드시 하나는 선택됨)

##### Step 6: 직관과 해석
- 동전 던지기는 2가지 결과(outcome) 중 하나 (앞/뒤) → 이항 분포
- 주사위 던지기는 6가지 결과(outcome) 중 하나 (1~6) → 다항 분포
- 분류 문제는 클래스 중 하나 선택 → 다항 분포!

- 비유: 선택지가 3개인 시험 문제 (고양이/개/새)
- 각각 선택될 확률: $p_1, p_2, p_3$ (합은 1)
- 이것이 다항 분포예요

##### Step 7: 조합 관점
- $\frac{n!}{n_1! \cdots n_k!}$: $n$개 중 $n_1$개, $n_2$개, ..., $n_k$개를 선택하는 방법의 수
- **예시**: 10번 시행에서 클래스 0이 3번, 클래스 1이 5번, 클래스 2가 2번 나오는 경우의 수
  - $\frac{10!}{3! \cdot 5! \cdot 2!} = \frac{3628800}{6 \cdot 120 \cdot 2} = 2520$가지 방법
- 이항 분포에서 $\binom{n}{k} = \frac{n!}{k!(n-k)!}$도 같은 원리!

##### Step 8: 통계량
- $E[X_i] = np_i$ (결과 $i$의 기대 횟수)
- $\text{Var}(X_i) = np_i(1-p_i)$
- $\text{Cov}(X_i, X_j) = -np_i p_j$ (다른 클래스가 많아지면 나머지는 감소)
- 예: 주사위를 60번 던지면 각 면의 기댓값은 10번, 분산은 약 8.33

##### Step 9: 머신러닝 연결
- 단일 샘플은 **카테고리 분포**($n=1$)를 따르고, 여러 샘플을 모으면 다항 분포
- 소프트맥스 출력은 카테고리 분포의 파라미터
- CrossEntropy Loss는 다항 분포에서 MLE를 적용한 결과, 로그우도 최대화와 동일

### 연속 확률 분포 (Continuous Probability Distribution)

#### Step 0: 용어 정리
- 연속 확률변수 $X$: 값이 실수 구간 전체를 가질 수 있음
- 확률밀도함수 $f(x)$: 구간을 적분하면 확률이 되는 함수

#### Step 1: 기본 공식
- $P(a \leq X \leq b) = \int_a^b f(x) dx$

#### 1) 정규 분포 (Normal Distribution)

##### Step 0: 파라미터 정리
- $\mu$: 평균 (분포의 중심)
- $\sigma^2$: 분산 (퍼진 정도)
- 표기: $X \sim \mathcal{N}(\mu, \sigma^2)$

##### Step 1: 확률밀도함수
$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

##### Step 2: 특징 요약
- 종 모양(bell curve) 형태
- 평균을 중심으로 좌우 대칭
- **68-95-99.7 규칙**: 평균 ±1σ(68%), ±2σ(95%), ±3σ(99.7%)
- 중심극한정리: 많은 독립 변수의 합 → 정규분포에 수렴

##### Step 3: 직관적 비유
- 키, 몸무게, 시험 점수 등 자연현상이 정규분포를 따르는 경우가 많음
- 평균 170cm라면 160~180cm가 가장 흔하고, 아주 작거나 큰 값은 드묾

##### Step 4: 머신러닝 연결
- **가중치 초기화**: Xavier, He 초기화는 정규분포 기반
- **회귀 오차 모델링**: 예측 오차가 정규분포라고 가정하면 MSE Loss 등장
- **배치 정규화**: 평균 0, 분산 1로 맞춰 학습 안정성 향상
