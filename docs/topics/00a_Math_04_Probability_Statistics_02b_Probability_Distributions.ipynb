**머신러닝에서의 의미**:

- 모델이 예측하는 값의 불확실성

### 이산 확률 분포 (Discrete Probability Distribution)

**이산이란?** 셀 수 있는 값들 (1, 2, 3, ... 또는 "고양이", "개", "새")

**확률 질량 함수 (Probability Mass Function, PMF)**: $P(X = x)$
- 각 값에 대해 정확히 그 값이 나올 확률

#### 1) 이항 분포 (Binomial Distribution)

**이항 분포란?** 같은 실험을 여러 번 반복했을 때, 성공 횟수의 확률 분포예요.

**베르누이 시행 (Bernoulli Trial)**:
- 한 번의 실험에서 **성공** 또는 **실패** 두 가지 결과만 있음
- 각 시행은 **독립적** (한 시행의 결과가 다른 시행에 영향 없음)
- 성공 확률 $p$는 모든 시행에서 동일

**예시**: 
- 동전 던지기: 앞면(성공, $p=0.5$) 또는 뒷면(실패, $1-p=0.5$)
- 시험 문제: 정답(성공) 또는 오답(실패)
- 제품 품질 검사: 합격(성공) 또는 불합격(실패)

**이항 분포**: 베르누이 시행을 $n$번 반복했을 때, 성공이 $k$번 나올 확률

$$P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}$$

여기서:
- $n$: 시행 횟수
- $k$: 성공 횟수 ($k = 0, 1, 2, ..., n$)
- $p$: 한 번의 시행에서 성공할 확률
- $1-p$: 한 번의 시행에서 실패할 확률

**공식 유도 과정**:

**단계 1: 경우의 수 구하기**

$n$번 시행에서 성공이 $k$번 나오는 경우의 수는?

**조합 (Combination)**: $\binom{n}{k} = \frac{n!}{k!(n-k)!}$

**의미**: $n$개 중에서 $k$개를 선택하는 방법의 수

**예시**: 5번 시행에서 성공이 3번 나오는 경우
- 성공 위치를 선택: $\binom{5}{3} = \frac{5!}{3!2!} = \frac{5 \times 4}{2 \times 1} = 10$가지
- (성공, 성공, 성공, 실패, 실패), (성공, 성공, 실패, 성공, 실패), ... 등 10가지

**단계 2: 각 경우의 확률 계산**

특정한 한 가지 경우를 생각해봅시다: 처음 $k$번이 성공, 나머지 $n-k$번이 실패

**독립 사건의 곱셈 원리**: 독립적인 사건들이 동시에 일어날 확률은 각 확률의 곱

$$P(\text{성공} \times k \text{번}, \text{실패} \times (n-k) \text{번}) = p \times p \times \cdots \times p \times (1-p) \times (1-p) \times \cdots \times (1-p)$$

$$= p^k \times (1-p)^{n-k}$$

**의미**:
- $p^k$: 성공이 $k$번 나올 확률
- $(1-p)^{n-k}$: 실패가 $(n-k)$번 나올 확률

**단계 3: 모든 경우의 확률 합산**

$n$번 시행에서 성공이 $k$번 나오는 방법은 $\binom{n}{k}$가지이고, 각 경우의 확률은 모두 $p^k (1-p)^{n-k}$로 같아요.

**합사건의 확률**: 서로 다른 경우들의 확률은 더하면 됩니다:

$$P(X = k) = \binom{n}{k} \times p^k (1-p)^{n-k}$$

**구체적인 계산 예시**:

**문제**: 공정한 동전을 5번 던질 때, 앞면이 정확히 3번 나올 확률은?

**해결**:
- $n = 5$, $k = 3$, $p = 0.5$

**조합 계산**:
$$\binom{5}{3} = \frac{5!}{3!2!} = \frac{5 \times 4 \times 3!}{3! \times 2 \times 1} = \frac{5 \times 4}{2 \times 1} = 10$$

**확률 계산**:
$$P(X = 3) = 10 \times (0.5)^3 \times (0.5)^2 = 10 \times 0.125 \times 0.25 = 10 \times 0.03125 = 0.3125$$

**답**: 약 31.25%의 확률로 앞면이 정확히 3번 나옵니다.

**기댓값과 분산**:

**기댓값 (Expected Value)**: $E[X] = np$

**직관적 이해**: 
- 한 번 시행에서 평균적으로 성공할 확률은 $p$
- $n$번 시행하면 평균적으로 $np$번 성공

**예시**: 동전을 100번 던질 때 ($n=100$, $p=0.5$)
- 기댓값: $E[X] = 100 \times 0.5 = 50$
- 앞면이 평균적으로 50번 나옵니다

**분산 (Variance)**: $\text{Var}(X) = np(1-p)$

**의미**: 분산이 클수록 성공 횟수가 평균에서 더 많이 퍼져있어요

**예시**: 동전을 100번 던질 때
- 분산: $\text{Var}(X) = 100 \times 0.5 \times 0.5 = 25$
- 표준편차: $\sqrt{25} = 5$

**머신러닝 연결**:
- 이항 분포는 **이진 분류 (Binary Classification)**의 기초
- 다항 분포는 이항 분포를 여러 결과(outcome/category)로 확장한 것

#### 2) 다항 분포 (Multinomial Distribution)

**다항 분포란?** 여러 개의 결과(outcome/category) 중 하나를 선택하는 확률 분포예요. 이항 분포를 여러 결과(outcome/category)로 확장한 것이에요.

**이항 분포와의 관계**:
- **이항 분포**: 2가지 결과(outcome) (성공/실패) → 다항 분포의 특수한 경우 ($k=2$)
- **다항 분포**: $k$개 결과(outcome/category) ($k \geq 2$)
- 이항 분포는 다항 분포에서 결과(outcome)가 2개일 때의 경우

**예시**:
- 동전 던지기: 2가지 결과(outcome) (앞/뒤) → **이항 분포**
- 주사위 던지기: 6가지 결과(outcome) (1~6) → **다항 분포**
- 분류 문제: 여러 클래스 중 하나 선택 → **다항 분포**

**다항 분포**: $n$번 시행에서 각 결과(outcome) $i$가 $n_i$번 나올 확률

$$P(X_1=n_1, X_2=n_2, \ldots, X_k=n_k) = \frac{n!}{n_1! n_2! \cdots n_k!} p_1^{n_1} p_2^{n_2} \cdots p_k^{n_k}$$

여기서:
- $n$: 전체 시행 횟수
- $n_i$: 결과(outcome) $i$가 나온 횟수 ($i = 1, 2, ..., k$)
- $p_i$: 한 번의 시행에서 결과(outcome) $i$가 나올 확률

**제약 조건**:

**시행 횟수 제약**: 
$$\sum_{i=1}^k n_i = n$$

**의미**: 모든 결과(outcome)가 나온 횟수의 합은 전체 시행 횟수와 같아야 해요

**확률 제약**:
$$\sum_{i=1}^k p_i = 1$$

**의미**: 모든 결과(outcome)의 확률을 더하면 1이 되어야 해요 (어떤 결과(outcome)든 반드시 하나는 선택됨)

**직관적 이해**: 
- 동전 던지기는 2가지 결과(outcome) 중 하나 (앞/뒤) → 이항 분포
- 주사위 던지기는 6가지 결과(outcome) 중 하나 (1~6) → 다항 분포
- 분류 문제는 클래스 중 하나 선택 → 다항 분포!

**비유**: 
- 선택지가 3개인 시험 문제 (고양이/개/새)
- 각각 선택될 확률: $p_1, p_2, p_3$ (합은 1)
- 이것이 다항 분포예요

**조합론적 의미**:
- $\frac{n!}{n_1! \cdots n_k!}$: $n$개 중 $n_1$개, $n_2$개, ..., $n_k$개를 선택하는 방법의 수
- **예시**: 10번 시행에서 클래스 0이 3번, 클래스 1이 5번, 클래스 2가 2번 나오는 경우의 수
  - $\frac{10!}{3! \cdot 5! \cdot 2!} = \frac{3628800}{6 \cdot 120 \cdot 2} = 2520$가지 방법
- 이항 분포에서 $\binom{n}{k} = \frac{n!}{k!(n-k)!}$도 같은 원리!

**기댓값과 분산**:

각 결과(outcome) $X_i$에 대해:

**기댓값 (Expected Value)**: $E[X_i] = np_i$

**의미**: 결과(outcome) $i$가 평균적으로 $np_i$번 나옵니다

**예시**: 주사위를 60번 던질 때 ($n=60$, $p_i = \frac{1}{6}$)
- 각 숫자가 나올 기댓값: $E[X_i] = 60 \times \frac{1}{6} = 10$
- 각 숫자가 평균적으로 10번 나옵니다

**분산 (Variance)**: $\text{Var}(X_i) = np_i(1-p_i)$

**의미**: 결과(outcome) $i$의 분산은 이항 분포와 같은 형태

**예시**: 주사위를 60번 던질 때
- 분산: $\text{Var}(X_i) = 60 \times \frac{1}{6} \times \frac{5}{6} = 60 \times \frac{5}{36} = \frac{300}{36} \approx 8.33$

**공분산 (Covariance)**: $\text{Cov}(X_i, X_j) = -np_i p_j$ (단, $i \neq j$)

**의미**: 한 결과(outcome)가 많이 나오면 다른 결과(outcome)는 덜 나오므로 음의 상관관계

**머신러닝 연결**: 
- **주의**: 엄밀히 말하면, 단일 데이터 포인트의 분류는 **카테고리 분포 (Categorical Distribution)**입니다
- 다항 분포는 $n$번 시행에서 각 카테고리가 나온 횟수의 분포이고, $n=1$일 때가 카테고리 분포입니다
- 분류 문제에서는 각 데이터 포인트가 독립적으로 카테고리 분포를 따르고, 여러 데이터를 모으면 다항 분포와 연결됩니다
- 소프트맥스 (Softmax) 출력이 카테고리 분포의 확률 파라미터
- CrossEntropy Loss가 이 가정에서 MLE로 유도됨

### 연속 확률 분포 (Continuous Probability Distribution)

**확률 밀도 함수 (Probability Density Function, PDF)**: $f(x)$, $P(a \leq X \leq b) = \int_a^b f(x) dx$

#### 1) 정규 분포 (Normal Distribution)

**정규 분포란?** 가장 많이 쓰이는 확률 분포예요. 종 모양(bell curve)으로 생겼어요.

$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

**파라미터**:
- $\mu$ (뮤): **평균 (Mean)** - 분포의 중심점
- $\sigma^2$ (시그마 제곱): **분산 (Variance)** - 퍼진 정도
- 표기: $X \sim \mathcal{N}(\mu, \sigma^2)$

**특징**:
- **종 모양 (bell curve)**: 가운데가 높고 양쪽으로 갈수록 낮아짐
- **평균 주변에 집중**: 대부분의 값이 평균 근처에 있어요
- **중심극한정리**: 많은 확률변수를 더하면 정규분포에 가까워짐
- **68-95-99.7 규칙**: 
  - 평균 ± 1σ 안에 68% 데이터
  - 평균 ± 2σ 안에 95% 데이터
  - 평균 ± 3σ 안에 99.7% 데이터

**비유**: 
- 키, 몸무게, 시험 점수 등 자연 현상들이 보통 정규분포를 따라요
- 평균 키가 170cm라면, 160cm나 180cm는 많지만, 150cm나 190cm는 드물죠

**머신러닝 연결**: 
- **가중치 초기화**: Xavier, He initialization은 정규분포 사용
- **회귀 문제의 오차**: 예측 오차가 정규분포를 따른다고 가정 → MSE Loss
- **배치 정규화 (Batch Normalization)**: 데이터를 정규분포로 만들기 (평균 0, 분산 1)
